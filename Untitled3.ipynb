```python
def simulate(processes, algorithm, **params):
    processes = deepcopy(processes)

    if algorithm == "FCFS":
        return fcfs(processes)

    elif algorithm == "SJF":
        return sjf(processes)

    elif algorithm == "SRTF":
        return srtf(processes)

    elif algorithm == "PRIORITY":
        return priority(processes)

    elif algorithm == "RR":
        quantum = params.get("quantum", 1)
        return round_robin(processes, quantum)

    else:
        raise ValueError(f"Неизвестный алгоритм: {algorithm}")
```

```python
def fcfs(processes):
    processes = sorted(
        processes,
        key=lambda p: (p["arrival"], p["pid"])
    )

    time = 0
    intervals = []

    for p in processes:
        if time < p["arrival"]:
            time = p["arrival"]

        start = time
        end = start + p["burst"]

        intervals.append({
            "pid": p["pid"],
            "start": start,
            "end": end
        })

        time = end

    return intervals
```

```python
def sjf(processes):
    processes = deepcopy(processes)

    time = 0
    intervals = []
    completed = set()

    while len(completed) < len(processes):

        available = [
            p for p in processes
            if p["arrival"] <= time and p["pid"] not in completed
        ]

        if not available:
            time = min(
                p["arrival"]
                for p in processes
                if p["pid"] not in completed
            )
            continue

        current = min(
            available,
            key=lambda p: (p["burst"], p["arrival"], p["pid"])
        )

        start = time
        end = start + current["burst"]

        intervals.append({
            "pid": current["pid"],
            "start": start,
            "end": end
        })

        time = end
        completed.add(current["pid"])

    return intervals
```

```python
def priority(processes):
    processes = deepcopy(processes)

    time = 0
    intervals = []

    remaining = {
        p["pid"]: p["burst"]
        for p in processes
    }

    completed = set()
    current_pid = None
    start_time = None

    while len(completed) < len(processes):

        available = [
            p for p in processes
            if p["arrival"] <= time
            and p["pid"] not in completed
            and remaining[p["pid"]] > 0
        ]

        if not available:
            time += 1
            continue

        current = min(
            available,
            key=lambda p: (
                p["priority"],
                p["arrival"],
                p["pid"]
            )
        )

        pid = current["pid"]

        if current_pid != pid:
            if current_pid is not None:
                intervals.append({
                    "pid": current_pid,
                    "start": start_time,
                    "end": time
                })

            current_pid = pid
            start_time = time

        remaining[pid] -= 1
        time += 1

        if remaining[pid] == 0:
            completed.add(pid)

    if current_pid is not None:
        intervals.append({
            "pid": current_pid,
            "start": start_time,
            "end": time
        })

    return intervals
```

```python
from collections import deque

def round_robin(processes, quantum):
    processes = deepcopy(processes)

    if quantum <= 0:
        raise ValueError("Quantum должен быть больше 0")

    processes.sort(key=lambda p: (p["arrival"], p["pid"]))

    remaining = {
        p["pid"]: p["burst"]
        for p in processes
    }

    ready_queue = deque()

    time = 0
    index = 0
    intervals = []

    while index < len(processes) or ready_queue:

        if not ready_queue:
            time = max(time, processes[index]["arrival"])

            while (
                index < len(processes)
                and processes[index]["arrival"] <= time
            ):
                ready_queue.append(processes[index])
                index += 1

        current = ready_queue.popleft()

        pid = current["pid"]

        run_time = min(
            quantum,
            remaining[pid]
        )

        start = time
        end = time + run_time

        intervals.append({
            "pid": pid,
            "start": start,
            "end": end
        })

        time = end
        remaining[pid] -= run_time

        while (
            index < len(processes)
            and processes[index]["arrival"] <= time
        ):
            ready_queue.append(processes[index])
            index += 1

        if remaining[pid] > 0:
            ready_queue.append(current)

    return intervals
```

```python

def calculate_metrics(processes, intervals):
    results = []

    for p in processes:
        pid = p["pid"]

        process_intervals = [
            interval for interval in intervals
            if interval["pid"] == pid
        ]

        if not process_intervals:
            continue

        first_start = min(i["start"] for i in process_intervals)
        completion = max(i["end"] for i in process_intervals)

        turnaround = completion - p["arrival"]
        waiting = turnaround - p["burst"]
        response = first_start - p["arrival"]

        results.append({
            "PID": pid,
            "Waiting": waiting,
            "Turnaround": turnaround,
            "Response": response
        })

    total_time = max(i["end"] for i in intervals) if intervals else 0
    busy_time = sum(i["end"] - i["start"] for i in intervals)

    cpu_utilization = (
        busy_time / total_time * 100
        if total_time > 0 else 0
    )

    return {
        "Average Waiting": sum(r["Waiting"] for r in results) / len(results),
        "Average Turnaround": sum(r["Turnaround"] for r in results) / len(results),
        "Average Response": sum(r["Response"] for r in results) / len(results),
        "CPU Utilization (%)": cpu_utilization,
        "Process Details": results
    }
```

```python

def srtf(processes):
    processes = deepcopy(processes)
    time = 0
    intervals = []
    remaining = {p["pid"]: p["burst"] for p in processes}
    completed = set()
    current_pid = None
    start_time = None

    while len(completed) < len(processes):
        available = [
            p for p in processes
            if p["arrival"] <= time
            and p["pid"] not in completed
            and remaining[p["pid"]] > 0
        ]

        if not available:
            if current_pid is not None:
                intervals.append({
                    "pid": current_pid,
                    "start": start_time,
                    "end": time
                })
                current_pid = None
                start_time = None

            time += 1
            continue

        current = min(
            available,
            key=lambda p: (
                remaining[p["pid"]],
                p["arrival"],
                p["pid"]
            )
        )

        pid = current["pid"]

        if current_pid != pid:
            if current_pid is not None:
                intervals.append({
                    "pid": current_pid,
                    "start": start_time,
                    "end": time
                })

            current_pid = pid
            start_time = time

        remaining[pid] -= 1
        time += 1

        if remaining[pid] == 0:
            completed.add(pid)

    if current_pid is not None:
        intervals.append({
            "pid": current_pid,
            "start": start_time,
            "end": time
        })

    return intervals
```

```python

def plot_schedule(intervals, title):
    fig, ax = plt.subplots(figsize=(12, 3))

    pids = list(dict.fromkeys(i["pid"] for i in intervals))
    positions = {pid: idx for idx, pid in enumerate(pids)}

    for interval in intervals:
        start = interval["start"]
        duration = interval["end"] - start
        pid = interval["pid"]

        ax.barh(
            positions[pid],
            duration,
            left=start,
            height=0.6
        )
        ax.text(
            start + duration / 2,
            positions[pid],
            pid,
            ha="center",
            va="center",
            fontsize=8
        )

    ax.set_yticks(list(positions.values()))
    ax.set_yticklabels(pids)
    ax.set_xlabel("Время")
    ax.set_title(title)
    ax.grid(axis="x", alpha=0.3)
    plt.tight_layout()
    plt.show()

plot_schedule(without_aging, "Priority без aging")
plot_schedule(with_aging, "Priority с aging")
```

```python

def validate_schedule(processes, intervals):
    burst = {p["pid"]: p["burst"] for p in processes}
    arrival = {p["pid"]: p["arrival"] for p in processes}
    executed = {pid: 0 for pid in burst}

    previous_end = 0

    for interval in intervals:
        pid = interval["pid"]
        start = interval["start"]
        end = interval["end"]

        if pid not in burst:
            return False, f"Неизвестный процесс: {pid}"

        if end <= start:
            return False, f"Некорректный интервал: {interval}"

        if start < arrival[pid]:
            return False, f"{pid} запущен раньше прибытия"

        if start < previous_end:
            return False, "Интервалы перекрываются или идут не по времени"

        executed[pid] += end - start
        previous_end = end

    for pid in burst:
        if executed[pid] != burst[pid]:
            return False, (
                f"{pid}: выполнено {executed[pid]}, "
                f"ожидалось {burst[pid]}"
            )

    return True, "Расписание корректно"


for name, intervals in [
    ("SRTF", srtf_intervals),
    ("RR q=3", rr3_intervals)
]:
    valid, message = validate_schedule(set1, intervals)
    print(name, "—", message)
```
